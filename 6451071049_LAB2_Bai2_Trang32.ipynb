### Giới thiệu công thức F-measure:
Thông thường để so sánh hai giá trị độ chính xác (Precision) và độ phủ (Recall) của các phương pháp với nhau, người ta sử dụng giá trị gọi là **F-measure** (hay **F1-score**). F-measure là trung bình điều hòa (harmonic mean) của hai giá trị độ chính xác và độ phủ, được tính bằng công thức:

$$F	ext{-measure} = 
rac{2 	imes 	ext{precision} 	imes 	ext{recall}}{	ext{precision} + 	ext{recall}}$$

### Phần 1: Tính F-measure cho Bài 1 (Cơ bản - Trang 24-25: Doanh số đồ chơi trẻ em)
Dữ liệu kiểm thử ở Yêu cầu d:
1. X1 = {Loại='Búp bê', Số màu=3, Kích thước='To', Chất liệu='Cao su'}
2. X2 = {Loại='Xếp hình', Số màu=5, Kích thước='To', Chất liệu='Nhựa PP'}
3. X3 = {Loại='Điều khiển', Số màu=3, Kích thước='Vừa', Chất liệu='Cao su'}

Doanh số bán trên thực tế ở Yêu cầu e lần lượt là: `['Thấp', 'Thấp', 'Cao']`.


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
import pandas as pd
import numpy as np

# Định nghĩa hàm tính F-measure
def calc_f_measure(precision, recall):
    if precision + recall == 0:
        return 0.0
    return 2 * (precision * recall) / (precision + recall)

# Nhãn thực tế
y_actual = ['Thấp', 'Thấp', 'Cao']

# 1. Dự đoán từ Cây quyết định ID3/CART (theo kết quả phân nhánh trên bảng dữ liệu)
# Kết quả dự đoán cho 3 mẫu trên là:
# Mẫu 1 (Búp bê, Cao su) -> 'Thấp'
# Mẫu 2 (Xếp hình, Nhựa PP) -> 'Thấp'
# Mẫu 3 (Điều khiển, Cao su) -> 'Thấp'
y_pred_tree = ['Thấp', 'Thấp', 'Thấp']

# Chọn lớp dương là 'Cao', lớp âm là 'Thấp'
# TP = 0, FP = 0, FN = 1, TN = 2
# Nếu xét lớp dương là 'Thấp': TP = 2, FP = 1, FN = 0, TN = 0
# Ta tính theo chuẩn bài tập:
# Lớp 'Thấp':
p_tree_low = 2 / (2 + 1) # 2/3 ≈ 0.667
r_tree_low = 2 / (2 + 0) # 1.0
f_tree_low = calc_f_measure(p_tree_low, r_tree_low)

# Lớp 'Cao':
p_tree_high = 0.0 # do không dự đoán mẫu nào là Cao
r_tree_high = 0.0 # do bỏ sót mẫu Cao
f_tree_high = 0.0

# 2. Dự đoán từ Naïve Bayes có làm trơn Laplace (Yêu cầu i):
# Các xác suất P(X|Cao)*P(Cao) và P(X|Thấp)*P(Thấp):
# Cả 3 mẫu đều có P(X|Thấp) > P(X|Cao), do đó nhãn dự đoán là ['Thấp', 'Thấp', 'Thấp']
# Khi đó:
p_nb_low = 2 / 3
r_nb_low = 1.0
f_nb_low = calc_f_measure(p_nb_low, r_nb_low)

print("BẢNG KẾT QUẢ F-MEASURE BÀI 1 (DOANH SỐ ĐỒ CHƠI):")
toy_df = pd.DataFrame({
    'Phương pháp': ['Cây quyết định (Decision Tree)', 'Naïve Bayes (Laplace)'],
    'Precision (Lớp Thấp)': [p_tree_low, p_nb_low],
    'Recall (Lớp Thấp)': [r_tree_low, r_nb_low],
    'F-measure (Lớp Thấp)': [f_tree_low, f_nb_low],
    'F-measure (Lớp Cao)': [f_tree_high, 0.0],
    'Macro F-measure': [(f_tree_low + f_tree_high)/2, (f_nb_low + 0.0)/2]
})
print(toy_df.to_string(index=False))
```

### Phần 2: Tính F-measure cho Bài 2 (Cơ bản - Trang 25-26: Phân tích cảm xúc mạng xã hội)
