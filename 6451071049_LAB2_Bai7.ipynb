### Bước 1: Chuẩn bị các thư viện cần thiết


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
import matplotlib
import matplotlib.pyplot as plt
from sklearn import datasets, tree, metrics
from sklearn.model_selection import train_test_split
from PIL import Image, ImageOps
import numpy as np

print("Đã import các thư viện thành công!")
```

### Bước 2: Tải dữ liệu có sẵn trong thư viện sklearn


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
# The digits dataset
digits = datasets.load_digits()
print("Kích thước tập ảnh digits:", digits.images.shape)
print("Số lượng nhãn tương ứng:", len(digits.target))
```

### Bước 3: Biểu diễn 4 mẫu dữ liệu huấn luyện tương ứng với 4 lớp 0, 1, 2, 3


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
_, axes = plt.subplots(1, 4, figsize=(10, 3))
images_and_labels = list(zip(digits.images, digits.target))
for ax, (image, label) in zip(axes, images_and_labels[:4]):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
    ax.set_title('Training: %i' % label)
plt.show()
```

### Bước 4: Chuyển đổi mỗi đối tượng hình ảnh (8x8) thành ma trận 1x64


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
# Flatten the image to turn data into a (samples, feature) matrix
n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))
print("Kích thước ma trận dữ liệu sau khi flatten:", data.shape)
```

### Bước 5: Khởi tạo mô hình cây quyết định ID3


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
# Create a classifier: a decision tree classifier
classifier = tree.DecisionTreeClassifier(criterion="entropy", random_state=33)
print("Mô hình khởi tạo:", classifier)
```

### Bước 6: Chia dữ liệu thành huấn luyện 80%, kiểm thử 20%


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
# Split data into train and test subsets
X_train, X_test, y_train, y_test = train_test_split(
    data, digits.target, test_size=0.2, shuffle=False)
print("Số lượng mẫu huấn luyện X_train:", len(X_train))
print("Số lượng mẫu kiểm thử X_test:", len(X_test))
```

### Bước 7: Tiến hành huấn luyện xây dựng cây ID3


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
# We learn the digits on the first part of the digits
classifier.fit(X_train, y_train)
print("Đã huấn luyện xong cây ID3!")
```

### Bước 8: Dự đoán nhãn dữ liệu cho tập kiểm thử


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
# Now predict the value of the digit on the second part:
predicted = classifier.predict(X_test)
print("5 nhãn dự đoán đầu tiên:", predicted[:5])
```

### Bước 9: Biểu diễn một vài kết quả dự đoán


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
_, axes = plt.subplots(1, 4, figsize=(10, 3))
images_and_predictions = list(zip(digits.images[n_samples // 2:], predicted))
for ax, (image, prediction) in zip(axes, images_and_predictions[:4]):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
    ax.set_title('Prediction: %i' % prediction)
plt.show()
```

### Bước 10: Thông số đạt được của cây ID3 và Ma trận nhầm lẫn


```python
print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
print("Classification report for classifier %s:\n%s\n"
      % (classifier, metrics.classification_report(y_test, predicted)))

cm = metrics.confusion_matrix(y_test, predicted)
print("Confusion matrix:\n", cm)

plt.figure(figsize=(8, 6))
disp = metrics.ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=classifier.classes_)
disp.plot(cmap='viridis', values_format='d')
plt.title("Confusion Matrix - ID3 Digit Classifier")
plt.show()
```

### Bước 11: Đọc và tiền xử lý hình ảnh chữ số viết tay thực tế
Đọc hình ảnh chữ số chụp bên ngoài (`dataset/test_image_Bai7 (1).jpg`), chuyển grayscale, resize về 8x8, đảo màu và phân ngưỡng nhị phân làm rõ nét ảnh.


```python
from google.colab import files
uploaded = files.upload()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageOps

print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)
img_path = 'test_image_Bai7.jpg'

# 1. Đọc và resize ảnh về 8x8 dạng grayscale
img = Image.open(img_path).convert("L").resize((8, 8))

# 2. Đảo màu ảnh để chữ trắng nền đen giống tập huấn luyện
img_inverted = ImageOps.invert(img)

# 3. Chuyển thành numpy array
im2arr = np.array(img_inverted)

# Hiển thị ảnh sau khi đảo màu
plt.figure(figsize=(4, 4))
plt.imshow(im2arr, cmap=plt.cm.gray_r, interpolation='nearest')
plt.title("Ảnh sau khi đảo màu và resize 8x8")
plt.show()

# 4. Chuyển từ ma trận 2D (8x8) thành 1D (1x64)
img1d = im2arr.reshape([1, 64])

# 5. Làm rõ nét ảnh bằng phân ngưỡng đen - trắng (ngưỡng 110)
img1d_norm = img1d.copy()
img1d_norm[img1d_norm > 109] = 155
img1d_norm[img1d_norm < 110] = 0

print("Mảng 1x64 sau khi làm rõ nét:\n", img1d_norm)

# Hiển thị ảnh sau khi phân ngưỡng nhị phân
plt.figure(figsize=(4, 4))
plt.imshow(img1d_norm.reshape(8, 8), cmap=plt.cm.gray_r, interpolation='nearest')
plt.title("Hình ảnh chữ số sau khi làm rõ (Binarized)")
plt.show()
```

### Bước 12: Tiến hành nhận diện bằng cây ID3 và in ra kết quả


```python
import numpy as np
from PIL import Image, ImageOps
from sklearn.datasets import load_digits
from sklearn import tree

print("Họ tên: Lương Thị Kim Ngân - MSSV: 6451071049\n" + "="*50)

# 1. Khởi tạo và huấn luyện classifier trên tập digits nếu chưa có trong kernel
digits = load_digits()
X_digits = digits.images.reshape((len(digits.images), -1))
y_digits = digits.target

classifier = tree.DecisionTreeClassifier(random_state=42)
classifier.fit(X_digits, y_digits)

# 2. Xử lý ảnh đầu vào test_image_Bai7.jpg
img = Image.open('test_image_Bai7.jpg').convert("L").resize((8, 8))
img_inv = ImageOps.invert(img)
img1d = np.array(img_inv).reshape([1, 64])

# Phân ngưỡng (ngưỡng 110) và chuẩn hóa về thang giá trị tập huấn luyện
img1d_norm = img1d.copy()
img1d_norm[img1d_norm > 109] = 16
img1d_norm[img1d_norm <= 109] = 0

# 3. Dự đoán nhãn chữ số viết tay
y_pred = classifier.predict(img1d_norm)
print("KẾT QUẢ DỰ ĐOÁN CHỮ SỐ VIẾT TAY TỪ HÌNH ẢNH:", y_pred)
print(f"-> Mô hình nhận diện chính xác hình ảnh là chữ số: {y_pred[0]}!")
```
